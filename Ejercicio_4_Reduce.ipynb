```python
from functools import reduce
#a. Dada una lista de numeros, calcula la multiplicación de estos.
numeros = [rep for rep in range(1,10)]
print(reduce(lambda x, y: x * y, numeros))
#b. Dada una lista de palabras, concetenalas en una sola.
palabras = ["radar", "Azul", "escuela","python", "ana", "oso", "java", "Luna", "isla", "php", "casa", "umbral"]

print(reduce(lambda x, y: x + "" + y, palabras))

#c. Dada una lista, hallar el mayor de todos, define una función auxiliar que dado dos numeros retorne el mayor

numeros = [3, 8, 15, 22, 7, 0, 41, 16, 9, 30, 5, 12]
def mayor(a,b):
    if a >= b:
        return a
    else:
        return b
print(reduce(lambda x, y: mayor(x,y), numeros))
```
