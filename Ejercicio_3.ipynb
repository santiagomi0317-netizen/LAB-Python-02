```python
#Realizar los siguientes ejercicios usando map.
#a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit.
def celsiusFaren(numero):
    return (numero *1.8) + 32

lista = [rep for rep in range(10)]
print(list(map(celsiusFaren, lista)))

#b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elementos (primer elemento de lista 1 + primer elemento lista 2).

lista1 = [rep for rep in range(10)]
lista2 = [rep for rep in range(10)]
print(list(map(lambda a,b: a+b, lista1,lista2)))

#c. Dada una lista de radios de circulos, generar la lista de sus areas.
radios = [rep for rep in range(10)]
print(list(map(lambda a: (a**2)*3.1416, radios)))
#d. Dada un diccionario, generar una lista de tuplas (nombre, valor+10%)
diccinario = {"pan": 1000, "leche": 2500, "café": 5000}

print(list(map(lambda par: (par[0], par[1]*1.10), diccinario.items())))
#e. Dada una matriz, multiplicar todos los elementos por 10.
def mult10(numero):
    return numero*10
matriz = [[1, 2, 3],[4, 5, 6],[7, 8, 9]]
print(list(map(lambda fila: list(map(mult10,fila)), matriz)))
```
