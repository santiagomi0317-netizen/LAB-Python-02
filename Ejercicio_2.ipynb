```python
#a. Generar una lista con tuplas (n, n^2) para una lista de numeros.
numeros = [1, 2, 3, 4, 5]
resultado = [(n, n**2) for n in numeros]
print(resultado)
#b. Generar una lista con cubos de los números mayores a 10.
resultado = [x**2  for x in range(20) if x > 10 ]
print(resultado)
#c. Generar una lista con solo los numeros pares.
resultado = [x  for x in range(20) if x % 2 == 0 ]
print(resultado)
#d. Dada la lista numeros, producir:
numeros = [1, 2, 3, 4, 5]
resultado = [("valor:", n*5) for n in numeros]
print(resultado)
#e. Generar una lista con la conversion correspondiente a grados Fahrenheit.
resultado = [(x*1.8) + 32  for x in range(10)]
print(resultado)
#f. Generar una lista filtrando las palabras de mas de 5 letras.
palabras = ["Azul", "casa", "Árbol", "perro", "avión", "Luna", "amigo", "fútbol", "arena", "Zebra"]
resultado = [x  for x in palabras if len(x)>5]
print(resultado)
#g. Generar una lista con solo las iniciales de las palabras.
palabras = ["Azul", "casa", "Árbol", "perro", "avión", "Luna", "amigo", "fútbol", "arena", "Zebra"]
resultado = [x[0]  for x in palabras]
print(resultado)
#h. Generar una lista indicando "par" o "impar" dada una lista de numeros.
numeros = [1, 2, 3, 4, 5]
resultado = [ "par" if numeros[rep] %2 == 0 else "impar" for rep in range(len(numeros))]
print(resultado)
#i. Dada una lista de listas (matriz), aplanala.
matriz=  [[1,2,3], [4,5,6], [7,8,9]]
resultado = [x for fila in matriz for x in fila]
print(resultado)

```
