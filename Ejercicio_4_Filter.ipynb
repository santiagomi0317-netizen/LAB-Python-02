```python
#a. Dada una lista de palabras, filtrar solo aquellas con mas de 4 letras.
palabras = ["python", "java", "javascript", "php", "perl"]

print(list(filter(lambda x:len(x)>5, palabras)))
#b. Dada una lista filtrar los elementos que son nulos (None).
palabras = ["python", "java", "javascript", "php", "perl", None]

print(list(filter(lambda x:x is None, palabras)))
#c. Dada una lista de palabras, filtrar aquellas que empiezan pro una vocal.
palabras = ["python", "java", "javascript", "php", "perl"]
def empiezaVocal(palabra):
    return palabra is not None and palabra.lower().startswith(("a", "e", "i", "o", "u"))
print(list(filter(empiezaVocal, palabras)))
#d. Dada una lista de palabras, filtrar aquellas que son palindromos.
palabras = ["python", "java", "javascript", "php", "perl"]
def palindromo(palabra):
    invertido = palabra[::-1]
    if palabra == invertido:
        return True
    else:
        return False
print(list(filter(palindromo, palabras)))
#e. Dada una lista de numeros, filtrar los que terminan en 5.
numeros = [x for x in range(10)]
def empiezaPor5(numero):
    palabra = str(numero)
    return palabra is not None and palabra.lower().endswith("5")
print(list(filter(empiezaPor5, numeros)))
#f. Dada una lista de diccionarios, filtrar los productos con precio mayor a 100.
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]

print(list(filter(lambda p: p["precio"] > 100, productos)))


```
