```python
#a. Ordenas una lista de palabras alfabeticamente.
n = ["hola", "como", "estas", "hoy"]
print(sorted(n))
#b. Ordenar una lista de tuplas por el primer elemento.
m = [(4, 1), (3, 5), (2, 6), (7, 8)]
print(sorted(m))
#c. Dada una lista de diccionarios, ordenar por precio (mayor a menor), y luego por nombre
productos = [
    ("Camisa", 30),
    ("Pantalón", 55),
    ("Medias", 10),
    ("Chaqueta", 80)
  ]
print(sorted(productos, key=lambda x: x[1], reverse=True))
print(sorted(productos))
```
